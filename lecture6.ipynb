```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import plotly.express as px

# Open the Excel file to inspect the actual sheet names
xls = pd.ExcelFile("Zomato_India_Orders_Dataset.xlsx")
print("Available sheets:", xls.sheet_names)

# Read the first available sheet or replace 'orders_Data' with the correct sheet name from the list above
correct_sheet = xls.sheet_names[0]
df = pd.read_excel("/content/Zomato_India_Orders_Dataset.xlsx", sheet_name=correct_sheet)
print(f"Data loaded from '{correct_sheet}': {df.shape[0]} orders and {df.shape[1]} variables")
```

```python
#1. Shape: tells us the number of rows and columns
print("Shape:", df.shape)

#2. Column: shows the names of all columns
print("\nColumns:")
print(df.columns)

#3. Index: shows the row labels
print("\nIndex:")
print(df.index)

#4. Data Types: tell us the type of data in each column
print("\nData Types:")
print(df.dtypes)
```

```python
#1.5 Classify the variables (we decide this using the data dictionary)
number_cols = ["Items_Count", "Order_Value", "Discount_Pct", "Delivery_Fee", "Distance_Km", "Delivery_Time_Min", "Rating", "Final_Amount"]
category_cols = ["City", "Cuisine", "Customer_Segment", "Platform", "Payment_Mode"]
date_cols = ["Order_Date"]
id_cols = ["Order_ID"]

print("Numerical:", len(number_cols), number_cols)
print("Categorical:", len(category_cols), category_cols)
print("Date:", date_cols)
print("ID:", id_cols)
```

```python
#1.6 Data quality check: missing values and duplicates
print("Missing values in each column:")
print(df.isnull().sum())

print("\nColumns with missing data:")
missing = df.isnull().sum()
print(missing[missing > 0])

print("\nPercent missing in rating:", round(df["Rating"].isnull().mean() * 100, 1), "%")

print("\nDuplicate rows: ", df.duplicated().sum())
print("Is Order_ID unique?:", df["Order_ID"].is_unique)
```

```python
#2.4 Quartiles and IQR (used later for outliers)

q1 = amount.quantile(0.25)
q2 = amount.quantile(0.5)
q3 = amount.quantile(0.75)
iqr = q3 - q1

print("Q1 (25% of orders paid less than)Rs", q1)
print("Q2 (median): Rs", q2)
print("Q3 (75% of orders paid less than) Rs", q3)
print("IQR (Q3 - Q1): Rs", iqr)
```

```python
#Quantile Quantile Plot for Normal DIstribution
# Step 1 : Import the libraries
import matplotlib.pyplot as plt
import scipy.stats as stats

# Step 2 : Remove missing values from Final_Amount
order_amount = df["Final_Amount"].dropna()

# Step 3 : Create the Q-Q plot
stats.probplot(order_amount, dist="norm", plot=plt)

# Step 4 : Add a title
plt.title("Q-Q Plot of final order Amount")

# Step 5 : Label the axes
plt.xlabel("Theoretical quantiles")

# Step 6 : Show the plot
plt.show()
```

```python
#Bar Plot

#Step 1: Find total revenue for each city
city_revenue = df.groupby("City")["Final_Amount"].sum()

#Step 2: Arrange cities from highest to lowest revenue
city_revenue = city_revenue.sort_values(ascending=False)

#Step 3: Print the revenue of each city
print(city_revenue)

#Step 4: Create a bar chart
city_revenue.plot(kind="bar")

#Step 5: Add chart title and y-axis label
plt.title("Total Revenue by City")
plt.ylabel("Total Revenue(₹)")

#Step 6: Show the chart
plt.show()
```

```python
# Step 1: Calculate total revenue for each cuisine
cuisine_revenue = df.groupby("Cuisine")["Final_Amount"].sum()

# Step 2: Average cuisine from highest to lowest revenue
cuisine_revenue = cuisine_revenue.sort_values(ascending=False)

# Step 3: Print the revenue of each cuisine
print(cuisine_revenue)

# Step 4: Create a bar chart
cuisine_revenue.plot(kind="bar")

# Step 5: Add chart title and y-axis label
plt.title("Revenue by Cuisine")
plt.ylabel("Total Revenue (Rs)")

# Step 6: Show the plot
plt.show()
```
